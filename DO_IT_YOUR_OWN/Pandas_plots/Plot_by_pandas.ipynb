# **Plots by using the Pandas Library**

## **What is Pandas Library?**
Pandas provides basic plotting functionality through its `.plot()` method, leveraging integration with Matplotlib to generate simple visualizations (e.g., line charts, bar plots). However, for advanced or customized plots, dedicated libraries like Seaborn or Matplotlib are typically preferred.

## **1. Define the term plot?**

A plot is a graphical representation of data, typically used to visualize relationships between variables, trends, or distributions. It is a visual tool used to communicate information, identify patterns, and make decisions. Plots can be used to display a wide range of data types, including numerical, categorical, and time-series data.

## **2. Why do we do ploting?**

We plot data to visualize the relationship between variables, to identify patterns and trends, to compare different groups , and to communicate results effectively. Plotting helps us to understand the data better and make informed decisions. 

## **3. Why plotting is important?**

Plotting is important because it helps us to:

- Identify patterns and trends in the data.
- Compare different groups or categories.
- Communicate results effectively to others.
- Make informed decisions based on the data.
- Identify outliers and anomalies in the data.
- Visualize complex data in a simple and intuitive way.

## **4. Pros and Cones of plots?**

**Pros of plots:**

- Helps to identify patterns and trends in the data.
- Communicates results effectively to others.
- Helps to make informed decisions based on the data.
- Visualizes complex data in a simple and intuitive way.
- Helps to identify outliers and anomalies in the data.

**Cons of plots:**

- Can be misleading if not created carefully.
- Can be difficult to interpret if not labeled correctly.
- Can be time-consuming to create if not using a tool or library.
- Can be difficult to compare different plots if not using a consistent scale or layout.

## **5. What are the things in Plots?**

- **X-axis (Horizontal)**: Represents the independent variable.  
- **Y-axis (Vertical)**: Represents the dependent variable.  
- **Labels**: Use `plt.xlabel("X-axis Label")` and `plt.ylabel("Y-axis Label")`.  
- **Scaling**: Adjust using `plt.xlim()` and `plt.ylim()`.  
- Set with `plt.title("Your Title")`.    
- Use `plt.legend(["Label1", "Label2"])`.  
- Add with `plt.grid(True)`.  
- Modify with `plt.xticks()` and `plt.yticks()`.  
- Rotate labels if needed using `rotation` parameter.    
- Customize appearance using:  
  - `linestyle="--"` (dashed lines)  
  - `marker="o"` (circle markers)  
  - `color="red"` (custom colors)  
- Create with `plt.subplot(rows, cols, index)`.  
- Useful for comparing data sets.  
- Highlight specific points using `plt.annotate("Text", xy=(x, y))`.  
- Use `plt.savefig("filename.png", dpi=300)` for high-quality images.  

## **6. What are the types of plots in Pandas library?**

- Line plot
- Bar plot
- Horizontal bar plot
- Area plot
- Scatter plot
- Box plot
- Hexbin plot
- Pie plot
- Histogram
- KDE plot

```python
# load the necessary library
import pandas as pd
```

```python
import seaborn as sns

# load the titanic dataset that is built in dataset
titanic=sns.load_dataset('titanic')
```

```python
titanic.head()
```

```python
# Create a horizontal bar plot
titanic.plot(kind='line', x='class', y='age')
```

```python
titanic.plot.hist(x='age', y='fare')
```

```python
# KDE plot with pandas library
titanic.plot.kde(x='age', y='fare')
```

```python
# Scatter plot with pandas library
titanic.plot.scatter(x='age', y='fare')
```

```python
# Pie chart with pandas library
titanic.groupby('class').size().plot(kind='pie')
```

```python
# Box plot with pandas library
titanic.boxplot(column='age', by='class')
```

```python
# Hexbin plot with pandas library
titanic.plot.hexbin(x='age', y='fare', gridsize=20)
```

```python
# Area plot with pandas library
titanic.plot.area(x='age', y='fare')
```

```python
# Bar plot with pandas library
titanic.groupby('class').size().plot(kind='bar')
```

```python
# Horizontal bar plot with pandas library
titanic.groupby('class').size().plot(kind='barh')
```
