# **Plots using the plotly Library.**

### **What is plotly library?**
The plotly library is a popular data visualization library in Python that allows users to create interactive,web based visualizations.It supports a wide range of chart types,including line plots,scatter plots,bar charts,histograms,box plots,and more.

## **1. Define the term plot?**

A plot is a graphical representation of data, typically used to visualize relationships between variables, trends, or distributions. It is a visual tool used to communicate information, identify patterns, and make decisions. Plots can be used to display a wide range of data types, including numerical, categorical, and time-series data.

## **2. Why do we do ploting?**

We plot data to visualize the relationship between variables, to identify patterns and trends, to compare different groups , and to communicate results effectively. Plotting helps us to understand the data better and make informed decisions. 

## **3. Why plotting is important?**

Plotting is important because it helps us to:

- Identify patterns and trends in the data.
- Compare different groups or categories.
- Communicate results effectively to others.
- Make informed decisions based on the data.
- Identify outliers and anomalies in the data.
- Visualize complex data in a simple and intuitive way.

## **4. Pros and Cones of plots?**

**Pros of plots:**

- Helps to identify patterns and trends in the data.
- Communicates results effectively to others.
- Helps to make informed decisions based on the data.
- Visualizes complex data in a simple and intuitive way.
- Helps to identify outliers and anomalies in the data.

**Cons of plots:**

- Can be misleading if not created carefully.
- Can be difficult to interpret if not labeled correctly.
- Can be time-consuming to create if not using a tool or library.
- Can be difficult to compare different plots if not using a consistent scale or layout.

## **5. What are the things in Plots?**

- **X-axis (Horizontal)**: Represents the independent variable.  
- **Y-axis (Vertical)**: Represents the dependent variable.  
- **Labels**: Use `plt.xlabel("X-axis Label")` and `plt.ylabel("Y-axis Label")`.  
- **Scaling**: Adjust using `plt.xlim()` and `plt.ylim()`.  
- Set with `plt.title("Your Title")`.    
- Use `plt.legend(["Label1", "Label2"])`.  
- Add with `plt.grid(True)`.  
- Modify with `plt.xticks()` and `plt.yticks()`.  
- Rotate labels if needed using `rotation` parameter.    
- Customize appearance using:  
  - `linestyle="--"` (dashed lines)  
  - `marker="o"` (circle markers)  
  - `color="red"` (custom colors)  
- Create with `plt.subplot(rows, cols, index)`.  
- Useful for comparing data sets.  
- Highlight specific points using `plt.annotate("Text", xy=(x, y))`.  
- Use `plt.savefig("filename.png", dpi=300)` for high-quality images.  

## **Interective Controls in Plotly**
Plotly provides several interactive controls that can be used to customize the appearance and behavior of a plot.
1. `Download plot as png`
You can download the plot as a PNG image by clicking on the "Download Plot" button in the top right corner of the plot.
2. `Zoom`
You can zoom in and out of the plot by using the mouse wheel or by clicking and dragging the plot with the mouse.
3. `Pan`
You can pan the plot by clicking and dragging the plot with the mouse.
4. `Box Select`
You can select a range of data points by clicking and dragging a box around them.
5. `Lasso Select`
You can select a range of data points by drawing a lasso around them.
6. `Auto Scale`
You can automatically scale the plot to fit the data by clicking on the "Auto Scale" button in the top right corner of the plot.
7. `Reset Axis`
You can reset the axis to their original state by clicking on the "Reset Axis" button in the top right corner of the plot.
8. `Toggle Spike Lines`
You can toggle the spike lines on and off by clicking on the "Toggle Spike Lines" button in the top right corner of the plot.
9. `Show Clossest Data on hover`
You can show the closest data point on hover by clicking on the "Show Clossest Data on hover" button in the top right corner of the plot.
10. `Compare Data on hover`
You can compare the data points on hover by clicking on the "Compare Data on hover" button in the top right corner of the plot.

## **Plotly Code Pattern**
Code pattern for creating a plots is to calling a plot function, passing a data form first argument, the x and y argument that  may be string and numeric and the axis on the x and y colomn is the x and y axis.That is as follows:

```python
px.plotly_fn(
    data = data,                     # data used in creating a plot
    x = x,                           # varaible x for the data
    y = y,                           # varaible y for the data
    title = title,                   # title of the plot
    xaxis_title = xaxis_title,       # title of the x axis
    yaxis_title = yaxis_title,       # title of the y axis
    width = width,                   # width of the plot
    height = height,                 # height of the plot
)
```

### **III. Financial & Time Series**

- Candlestick charts
- OHLC charts (Open-High-Low-Close)
- Time series sliders/range selectors

````python
# Create a time series plot with range slider and selector
fig = px.line(titanic, x='age', y='fare', title='Fare by Age with Range Slider and Selector')

# Update layout to include range slider and selector
fig.update_layout(
    xaxis=dict(
        rangeselector=dict(
            buttons=list([
                dict(count=1, label="1m", step="month", stepmode="backward"),
                dict(count=6, label="6m", step="month", stepmode="backward"),
                dict(step="all")
            ])
        ),
        rangeslider=dict(visible=True),
        type="linear"
    )
)

fig.show()
````

````python
# Create a sample dataset for OHLC chart
ohlc_data = pd.DataFrame({
    'date': pd.date_range(start='2023-01-01', periods=10, freq='D'),
    'open': np.random.rand(10) * 100,
    'high': np.random.rand(10) * 100 + 50,
    'low': np.random.rand(10) * 100,
    'close': np.random.rand(10) * 100
})

# Create an OHLC chart using the plotly library
fig = go.Figure(data=[go.Ohlc(x=ohlc_data['date'],
                              open=ohlc_data['open'],
                              high=ohlc_data['high'],
                              low=ohlc_data['low'],
                              close=ohlc_data['close'])])

# Update layout
fig.update_layout(title='OHLC Chart', xaxis_title='Date', yaxis_title='Value')

# Show the plot
fig.show()
````

````python
# Create a candlestick plot using the plotly library
fig = go.Figure(data=[go.Candlestick(x=ohlc_data['date'],
                                     open=ohlc_data['open'],
                                     high=ohlc_data['high'],
                                     low=ohlc_data['low'],
                                     close=ohlc_data['close'])])

# Update layout
fig.update_layout(title='Candlestick Chart', xaxis_title='Date', yaxis_title='Value')

# Show the plot
fig.show()
````

## **Customizing Plots in Plotly**
The code pattern for customizing the plot is to save the figure object returned from plotting function call it, `update_traces() function` and than show the `show() function`.

````python
# create a plot using the plotly library, create a plot and than use the update traces function and than show the plot
fig = px.scatter(titanic, x='age', y='fare', color='class')
fig.update_traces(mode='markers')           # update the mode of the plot to 'markers'
fig.show()
````

## **Customizing markers in plotly**
When wroking the visualization like scatter or line plot we can customize the markers according to the certain properties:
1. Size (set the marker size)
2. Color (set the marker color)
3. Symbol (set the marker symbol)
4. Opacity (set the marker transprancy)
5. Line (set the marker line properties)

````python
# The example of customizing the markers in plotly 
plot = px.scatter(titanic, x='age', y='fare', color='class')
plot.update_traces(marker=dict(size=12,
                    symbol='cross',
                   opacity=0.8,
                   line=dict(width=2,color='DarkSlateGrey')),
                   selector=dict(mode='markers')
                   )
plot.show()
````

## **Customizing Lines in Plots**
When working with visualizaion that contain certain lines we can do the certain things that are as follows:
1. Color (set the line color.)
2. Dash (select the dash style like `solid`, `dot` , `dash` , `longdash`, `dashdot`, `longdashdot`)
3. Shape (set how values are connected `linear`, `spline`, `hv`, `vh`, `hvh`,`vhv`.)
4. Width (set the line width.)

````python
# now the example for customizing lines in the graph 
plot_1 = px.line(titanic, x='age', y='fare', color='class')
plot_1.update_traces(line=dict(width=4, dash='dot', color='blue'), selector=dict(mode='lines'))
plot_1.show()
````

## **Customizing Bars in plots**
When working with barplots and histogram, you can update the bar themselves according to following properties:
1. Size (set the marker size)
2. Color (set the marker color)
3. Opacity (set the marker transprancy)
4. Line (the the width and color of border)
5. Symbol (set the shape of symbol)

````python
# now give me the example of the custoimizing the  bars in the plotly
plot_2 = px.bar(titanic, x='class', y='fare', color='class')
plot_2.update_traces(marker_color='green', selector=dict(mode='markers'))
plot_2.show()
````

````python
# now give me the example of the customizing of the histogram in the plotly
plot_3 = px.histogram(titanic, x='age', color='class')
plot_3.update_traces(marker_color='purple', selector=dict(mode='markers'))
plot_3.show()
````
