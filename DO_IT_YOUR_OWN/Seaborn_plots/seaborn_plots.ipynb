# **Plot by using the Seabron library**

## 1. What is Seabron Library?
Seaborn is a Python data visualization library based on Matplotlib that provides a high-level interface for creating statistically informative and aesthetically appealing graphs, such as heatmaps, distribution plots, and regression visualizations, with seamless integration into pandas DataFrames.

## **1. Define the term plot?**

A plot is a graphical representation of data, typically used to visualize relationships between variables, trends, or distributions. It is a visual tool used to communicate information, identify patterns, and make decisions. Plots can be used to display a wide range of data types, including numerical, categorical, and time-series data.

## **2. Why do we do ploting?**

We plot data to visualize the relationship between variables, to identify patterns and trends, to compare different groups , and to communicate results effectively. Plotting helps us to understand the data better and make informed decisions. 

## **3. Why plotting is important?**

Plotting is important because it helps us to:

- Identify patterns and trends in the data.
- Compare different groups or categories.
- Communicate results effectively to others.
- Make informed decisions based on the data.
- Identify outliers and anomalies in the data.
- Visualize complex data in a simple and intuitive way.

## **4. Pros and Cones of plots?**

**Pros of plots:**

- Helps to identify patterns and trends in the data.
- Communicates results effectively to others.
- Helps to make informed decisions based on the data.
- Visualizes complex data in a simple and intuitive way.
- Helps to identify outliers and anomalies in the data.

**Cons of plots:**

- Can be misleading if not created carefully.
- Can be difficult to interpret if not labeled correctly.
- Can be time-consuming to create if not using a tool or library.
- Can be difficult to compare different plots if not using a consistent scale or layout.

## **5. What are the things in Plots?**

- **X-axis (Horizontal)**: Represents the independent variable.  
- **Y-axis (Vertical)**: Represents the dependent variable.  
- **Labels**: Use `plt.xlabel("X-axis Label")` and `plt.ylabel("Y-axis Label")`.  
- **Scaling**: Adjust using `plt.xlim()` and `plt.ylim()`.  
- Set with `plt.title("Your Title")`.    
- Use `plt.legend(["Label1", "Label2"])`.  
- Add with `plt.grid(True)`.  
- Modify with `plt.xticks()` and `plt.yticks()`.  
- Rotate labels if needed using `rotation` parameter.    
- Customize appearance using:  
  - `linestyle="--"` (dashed lines)  
  - `marker="o"` (circle markers)  
  - `color="red"` (custom colors)  
- Create with `plt.subplot(rows, cols, index)`.  
- Useful for comparing data sets.  
- Highlight specific points using `plt.annotate("Text", xy=(x, y))`.  
- Use `plt.savefig("filename.png", dpi=300)` for high-quality images.  

## **6. What are the types of plots in Pandas library?**

Seaborn supports a variety of statistical and categorical plots, including:

- **Distribution plots** (e.g., `histplot, kdeplot, ecdfplot, displot`).
- **Categorical plots** (e.g., `barplot, boxplot, violinplot, stripplot, swarmplot, catplot`).
- **Relational plots** (e.g., `scatterplot, lineplot, relplot`).
- **Regression plots** (e.g., `lmplot, regplot`).
- **Matrix/Grid plots** (e.g., `heatmap, clustermap, pairplot, jointplot`).
- **Time series and multivariate visualizations** (e.g.,` using lineplot or FacetGrid`).

```python
# Import the seaborn library 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the example tips dataset
tips = sns.load_dataset("tips")
tips.head()
```

### **I.Distribution Plots**
1. Histogram.
2. Kdeplot.
3. ECDF Plot.
4. Displot.

```python
# Create a Histogram using Seaborn Library
sns.histplot(tips['total_bill'])
plt.show()
```

```python
# Create a KDE plot using Seaborn Library
sns.kdeplot(tips['total_bill'])
plt.show()
```

```python
# Create ECDF plot using Seaborn Library
sns.ecdfplot(tips['size'])
plt.show()
```

```python
# Create Distplot using Seaborn Library
sns.distplot(tips['tip'])
plt.show()
```

### **II. Categorical Plots**
1. Boxplot.
2. Violin Plot.
3. Bar Plot.
4. Swamp plot.
5. Stripplot.
6. Catplot.

```python
# Create a Bxo plot using Seaborn Library
sns.boxplot(x='day', y='total_bill', data=tips)
plt.show()
```

```python
# Create a Violin plot using Seaborn Library
sns.violinplot(x='day', y='total_bill', data=tips)
plt.show()
```

```python
# Create a Bar plot using Seaborn Library
sns.barplot(x='day', y='total_bill', data=tips)
plt.show()
```

```python
# Create a Swamp plot using Seaborn Library
sns.swarmplot(x='day', y='total_bill', data=tips)
plt.show()
```

```python
# Strip plot using Seaborn Library
sns.stripplot(x='day', y='total_bill', data=tips)
plt.show()
```

```python
# Create a Cat plot using Seaborn Library
sns.catplot(x="day", y="total_bill", hue="sex", kind="swarm", data=tips)
plt.show()
```

### **III. Relational Plots**
1. Scatter plot.
2. Line plot.
3. Rel plot.

```python
# Create a Scatter plot using Seaborn Library
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.show()
```

```python
# Create a Line plot using Seaborn Library
sns.lineplot(x='total_bill', y='tip', data=tips)
plt.show()
```

```python
# Create a Relplot using Seaborn Library
sns.relplot(x='total_bill', y='tip', data=tips)
plt.show()
```

### **IV. Regression Plots**
1. Reg plots.
2. lm plots.

```python
# Create a Regplot using Seaborn Library
sns.regplot(x='total_bill', y='tip', data=tips)
plt.show()
```

```python
# Create a Lmplot using Seaborn Library
sns.lmplot(x='total_bill', y='tip', data=tips)
plt.show()
```

### **V. Matrix/Grid plots.**
1. heatmap.
2. clustermap.
3. pairplot.
4. jointplot.

```python
# Create a Heatmap using Seaborn Library
sns.heatmap(tips.corr(), annot=True, cmap='coolwarm')
plt.show()
```

```python
# Create a Heatmap using Seaborn Library
sns.heatmap(tips[['size', 'total_bill', 'tip']].corr())
plt.show()
```

```python
### Create a clustermap using Seaborn Library
sns.clustermap(tips[['size', 'total_bill', 'tip']].corr())
plt.show()
```

```python
### Create a Jointplot using Seaborn Library
sns.jointplot(x='total_bill', y='tip', data=tips)
plt.show()
```

```python
### Create a Pairplot using Seaborn Library
sns.pairplot(tips)      # Pairplot of all the numerical columns
plt.show()
```

### **VI.Time series and multivariate visualizations**
1. lineplot. 
2. FacetGrid.

```python
# Create a lineplot using Seaborn Library
sns.lineplot(x='total_bill', y='tip', data=tips)
plt.show()
```

```python
# Create a facegrid using Seaborn Library
g = sns.FacetGrid(tips, col="time")
g.map(plt.scatter, "total_bill", "tip")
plt.show()
```
