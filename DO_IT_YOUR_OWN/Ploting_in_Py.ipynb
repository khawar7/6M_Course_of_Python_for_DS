# **Ploting in Python on an online dataset.**

## What is plotting?
Plotting is the process of creating a visual representation of data, often using a graph or chart.Plotting can be used to display data in a way that is easy to understand and interpret, making it a useful tool for communication and decision-making.

## What is a plot?
A plot is a graphical representation of data, often used to display the relationship between variables.It can be a line graph, bar chart, scatter plot, or other types of charts.

## What is a plot in Python?
In Python, a plot is a graphical representation of data created using a library such as Matplotlib or Seaborn.
