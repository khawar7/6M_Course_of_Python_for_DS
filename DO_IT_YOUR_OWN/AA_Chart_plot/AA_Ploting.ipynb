### **I. Distribution in the dataset.**

A distribution in statistics describes how the values of a variable are spread across a dataset. It specifies the possible values and their frequencies (or probabilities), providing insights into central tendency (e.g., mean, median), variability (e.g., range, variance), shape (e.g., symmetry, skewness), modality (e.g., unimodal, bimodal), and the presence of outliers. Distributions can be empirical (observed in data) or theoretical (e.g., normal, uniform).

To check the Distribution in the dataset using plots we have three different plots:

i. One Variable (`Few Data Points Colomn Histogram and Line Histogram`).\
ii. Two Variables (`Scatter Chart`).\
iii. Three Variables (`3D Area Chart`).

## **II. Relationhips**

In statistics and data analysis, a relationship refers to the association, connection, or pattern between two or more variables. It describes how changes in one variable correspond to changes in another, allowing us to identify dependencies, trends, or interactions in a dataset. Relationships can be:

- Quantitative (e.g., numerical correlations) or qualitative (e.g., categorical associations).
- Linear (following a straight-line pattern) or non-linear (e.g., exponential, quadratic).
- Positive (both variables increase together) or negative (one increases as the other decreases).
- Causal (one variable directly influences another) or non-causal (linked by coincidence or confounding factors).

To check the realtionship between the variables in the dataset we have two type of plots:

i. Scatter plot(two varaibles).\
ii. Bubble Chart (three variables).

## **III. Comparison**

In statistics, comparison refers to the process of evaluating differences or similarities between groups, categories, or variables in a dataset. When visualized through plots, comparisons highlight how distributions, trends, or summary statistics (e.g., means, medians) differ across predefined segments of data. The goal is to identify patterns, outliers, or relationships that inform decision-making or hypotheses.

### **Key Elements to Analyze in Comparative Plots**

1. Overlap vs. Separation:

- Overlapping distributions (e.g., histograms) suggest similar groups.
- Clear separation (e.g., non-overlapping box plots) indicates significant differences.

2. Trend Direction:

- Parallel lines (line plots) = consistent trends.
- Crossing lines = divergent outcomes.

3. Statistical Significance:

Use error bars (bar plots) or confidence intervals to infer significance.

To check the comparison of variables in the anscombe dataset, we use the following plots:

- One Variable per Item

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Load the built-in diamonds dataset
diamonds = sns.load_dataset("diamonds")

# Prepare data: Group by "cut" and calculate mean price
data = diamonds.groupby("cut")["price"].mean().reset_index()
categories = data["cut"].tolist()
values = data["price"].tolist()

# Set up angles and polar plot
theta = np.linspace(0.0, 2 * np.pi, len(categories), endpoint=False)  # Angles for each category
theta = np.concatenate((theta, [theta[0]]))  # Close the circle
values = np.concatenate((values, [values[0]]))  # Close the data

# Create the plot
fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(111, polar=True)

# Plot the filled area
ax.fill(theta, values, color="skyblue", alpha=0.5, label="Average Price")

# Customize labels and title
ax.set_xticks(theta[:-1])
ax.set_xticklabels(categories)
ax.set_title("Circular Area Plot: Average Diamond Price by Cut", pad=20, fontsize=14)
ax.legend(loc="upper right")

# Rotate the plot to start at the top (12 o'clock)
ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)  # Clockwise

plt.show()
```

### **IV. Composition**

In statistics, composition refers to the relative proportions or contributions of different components (subgroups, categories, or variables) that make up a whole. It focuses on how parts of a dataset or population are structured in relation to each other, often expressed as percentages, ratios, or fractions that sum to a constant total (e.g., 100%).

**Key Characteristics**

1. `Additivity`: All components sum to a fixed total (e.g., percentages of a budget allocated to departments).
2. `Relativity`: Components are analyzed in relation to each other, not as absolute values.
3. `Constrained Data`: Compositional data lies in a simplex space (e.g., data where each observation is a set of non-negative values summing to 1).

To check the compistion of the variables in the dataset we use the following plots that are:

i. Changing over time

- Few Periods 
1. Only relative differences matters (Stacked 100% Colomn Chart).
2. Relative and absolute differences matters (Stacked Colomn Chart).

- Many Periods
1. Only relative differences matters (Stacked 100% Colomn Chart).
2. Relative and absolute differences matters (Stacked Colomn Chart).

ii. Static

- Simple Share of Total (Pie Chart).
- Accomulation or Substraction of total (Water fall Chart).
- Components of Components (Stacked 100% Colomn Chart with Subcomponents).

```python
# Create a pivot table to summarize the data
pivot_table = anscombe.pivot_table(values='price', index='cut', columns='color', aggfunc=np.mean)

# Normalize the pivot table to get percentages
pivot_table_percentage = pivot_table.div(pivot_table.sum(axis=1), axis=0) * 100

# Create a stacked 100% column chart
pivot_table_percentage.plot(kind='bar', stacked=True, figsize=(10, 6), colormap='viridis')

# Labels and title
plt.xlabel("Cut")
plt.ylabel("Percentage")
plt.title("Stacked 100% Column Chart of Price by Cut and Color")

# Show the plot
plt.show()
```

```python
# Create a stacked column chart
pivot_table.plot(kind='bar', stacked=True, figsize=(10, 6), colormap='viridis')

# Labels and title
plt.xlabel("Cut")
plt.ylabel("Average Price")
plt.title("Stacked Column Chart of Average Price by Cut and Color")

# Show the plot
plt.show()
```

```python
# Normalize the pivot table to get percentages
pivot_table_percentage = pivot_table.div(pivot_table.sum(axis=1), axis=0) * 100

# Create a stacked 100% area chart
pivot_table_percentage.plot(kind='area', stacked=True, figsize=(10, 6), colormap='viridis')

# Labels and title
plt.xlabel("Cut")
plt.ylabel("Percentage")
plt.title("Stacked 100% Area Chart of Price by Cut and Color")

# Show the plot
plt.show()
```

```python
import plotly.graph_objects as go

# Sample built-in dataset (manually defined)
data = {
    'Category': ['Start', 'Sales', 'Expenses', 'Tax', 'Profit'],
    'Value': [1000, 500, -200, -100, 1200],  # Start value, changes, and final total
    'Measure': ['absolute', 'relative', 'relative', 'relative', 'total']
}

# Create Waterfall chart
fig = go.Figure(go.Waterfall(
    x=data['Category'],
    y=data['Value'],
    measure=data['Measure']
))

# Update layout
fig.update_layout(
    title="Business Revenue Waterfall Chart",
    yaxis_title="Amount ($)",
    xaxis_title="Stages"
)

# Show the chart
fig.show()
```
