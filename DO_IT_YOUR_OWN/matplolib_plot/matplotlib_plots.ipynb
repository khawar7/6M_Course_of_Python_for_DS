# **Plots using matplotlib Library**

## **What is matplotlib library?**

Matplotlib is a plotting library for the Python programming language and its numerical mathematics extension NumPy. It
provides an object-oriented API for embedding plots into applications using general-purpose GUI toolkits like Tkinter,wxPython,Qt, or GTK+,and for producing stand-alone plots suitable for saving as PDF,PNG, EPS files or displaying on the screen Matplotlib is a widely used library in the Python data science ecosystem,and it is often used in combination with other libraries like NumPy, Pandas, and Scikit-learn.

## **1. Define the term plot?**

A plot is a graphical representation of data, typically used to visualize relationships between variables, trends, or distributions. It is a visual tool used to communicate information, identify patterns, and make decisions. Plots can be used to display a wide range of data types, including numerical, categorical, and time-series data.

## **2. Why do we do ploting?**

We plot data to visualize the relationship between variables, to identify patterns and trends, to compare different groups , and to communicate results effectively. Plotting helps us to understand the data better and make informed decisions. 

## **3. Why plotting is important?**

Plotting is important because it helps us to:

- Identify patterns and trends in the data.
- Compare different groups or categories.
- Communicate results effectively to others.
- Make informed decisions based on the data.
- Identify outliers and anomalies in the data.
- Visualize complex data in a simple and intuitive way.

## **4. Pros and Cones of plots?**

**Pros of plots:**

- Helps to identify patterns and trends in the data.
- Communicates results effectively to others.
- Helps to make informed decisions based on the data.
- Visualizes complex data in a simple and intuitive way.
- Helps to identify outliers and anomalies in the data.

**Cons of plots:**

- Can be misleading if not created carefully.
- Can be difficult to interpret if not labeled correctly.
- Can be time-consuming to create if not using a tool or library.
- Can be difficult to compare different plots if not using a consistent scale or layout.

## **5. What are the things in Plots?**

- **X-axis (Horizontal)**: Represents the independent variable.  
- **Y-axis (Vertical)**: Represents the dependent variable.  
- **Labels**: Use `plt.xlabel("X-axis Label")` and `plt.ylabel("Y-axis Label")`.  
- **Scaling**: Adjust using `plt.xlim()` and `plt.ylim()`.  
- Set with `plt.title("Your Title")`.    
- Use `plt.legend(["Label1", "Label2"])`.  
- Add with `plt.grid(True)`.  
- Modify with `plt.xticks()` and `plt.yticks()`.  
- Rotate labels if needed using `rotation` parameter.    
- Customize appearance using:  
  - `linestyle="--"` (dashed lines)  
  - `marker="o"` (circle markers)  
  - `color="red"` (custom colors)  
- Create with `plt.subplot(rows, cols, index)`.  
- Useful for comparing data sets.  
- Highlight specific points using `plt.annotate("Text", xy=(x, y))`.  
- Use `plt.savefig("filename.png", dpi=300)` for high-quality images.

```python
# import the necessary libraries    
import matplotlib.pyplot as plt
```

```python
# import the seaborn library for built-in dataset
import seaborn as sns
```

```python
# load the iris dataset
iris = sns.load_dataset('iris')
iris.head()
```

```python
# Create a line plot using matplotlib
plt.plot(iris['sepal_length'], iris['sepal_width'])
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
```

```python
# Create a Bar plot using matplotlib
plt.bar(iris['species'], iris['sepal_length'])
plt.xlabel('Species')
plt.ylabel('Sepal Length')
plt.show()
```

```python
# Create a Scatter plot using matplotlib
plt.scatter(iris['sepal_length'], iris['sepal_width'])
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
```

```python
# Create a Histogram using matplotlib
plt.hist(iris['sepal_length'], bins=30)
plt.xlabel('Sepal Length')
plt.ylabel('Frequency')
plt.show()
```

```python
# Create a Box plot using matplotlib
plt.boxplot(iris['sepal_length'])
plt.ylabel('Sepal Length')
plt.show()
```

```python
# Create a Violin plot using seaborn
sns.violinplot(x='species', y='sepal_length', data=iris)
plt.show()
```

```python
# Create a error plot using matplotlib
plt.errorbar(iris['sepal_length'], iris['sepal_width'], xerr=0.1, yerr=0.1, fmt='o')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
```

```python
# Calculate the percentage of each species
species_counts = iris['species'].value_counts()
species_percentage = species_counts / species_counts.sum() * 100

# Create a pie chart
plt.pie(species_counts, labels=species_counts.index, autopct='%1.1f%%')
plt.title('Species Distribution in Iris Dataset')
plt.show()
```

```python
# Create a Stem plot using matplotlib
plt.stem(iris['sepal_length'], iris['sepal_width'])
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
```

```python
# Create a Stack plot using matplotlib
plt.stackplot(iris['sepal_length'], iris['sepal_width'])
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
```

```python
# Create a polor plot using matplotlib
plt.polar(iris['sepal_length'], iris['sepal_width'])
plt.show()
```

```python
# Create a 3D plot using matplotlib
from mpl_toolkits.mplot3d import Axes3D
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(iris['sepal_length'], iris['sepal_width'], iris['petal_length'])
ax.set_xlabel('Sepal Length')
ax.set_ylabel('Sepal Width')
ax.set_zlabel('Petal Length')
plt.show()
```

```python
# Compute the correlation matrix for specific columns
corr_matrix = iris[['petal_length', 'petal_width']].corr()

# Create an imshow plot
plt.imshow(corr_matrix, cmap='hot', interpolation='nearest')
plt.colorbar()
plt.show()
```

```python
from scipy.stats import kde
import numpy as np

# Select two features for contour plotting
x = iris["petal_length"]
y = iris["petal_width"]

# Create a 2D density estimation
k = kde.gaussian_kde([x, y])
xi, yi = np.meshgrid(np.linspace(x.min(), x.max(), 100),
                     np.linspace(y.min(), y.max(), 100))
zi = k(np.vstack([xi.flatten(), yi.flatten()]))

# Create the contour plot
plt.figure(figsize=(8, 6))
plt.contourf(xi, yi, zi.reshape(xi.shape), cmap="coolwarm", levels=20)
plt.colorbar(label="Density")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Contour Plot of Petal Length vs Petal Width")
plt.scatter(x, y, c="black", s=10, alpha=0.5)  # Add scatter plot for reference
plt.show()
```

```python
# Create a subplot with 2 rows and 2 columns
fig, axs = plt.subplots(2, 2, figsize=(10, 10))

# Plot the first subplot
axs[0, 0].plot(iris['sepal_length'], iris['sepal_width'])
axs[0, 0].set_title('Sepal Length vs Sepal Width')

# Plot the second subplot
axs[0, 1].scatter(iris['sepal_length'], iris['sepal_width'])
axs[0, 1].set_title('Sepal Length vs Sepal Width')

# Plot the third subplot
axs[1, 0].hist(iris['sepal_length'], bins=30)
axs[1, 0].set_title('Histogram of Sepal Length')

# Plot the fourth subplot
axs[1, 1].boxplot(iris['sepal_length'])
axs[1, 1].set_title('Boxplot of Sepal Length')

# Adjust spacing between subplots
plt.tight_layout()

# Show the plot
plt.show()
```

```python
#  Create a annotation plot using matplotlib
plt.plot(iris['sepal_length'], iris['sepal_width'])
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
# Add annotation
plt.annotate('This is an annotation', xy=(5, 5), xytext=(10, 10), arrowprops=dict(facecolor='red', shrink=0.05))
plt.show()
```
