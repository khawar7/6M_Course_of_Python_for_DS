# **Matplotlib CheetSheet**

Matplotlib is a comprehensive Python library for creating high-quality, customizable visualizations in data analysis and scientific computing. It provides a flexible framework for generating a wide range of plots, including line charts, scatter plots, bar graphs, and heatmaps, suitable for both static and interactive applications. Built on NumPy, it integrates seamlessly with other scientific Python libraries like Pandas and SciPy. Its object-oriented API allows precise control over plot elements, enabling users to tailor aesthetics and layouts to specific needs. Matplotlib supports various backends for rendering plots in different formats, such as PNG, PDF, and SVG, ensuring compatibility across platforms. Widely used in academia and industry, it remains a cornerstone for data visualization in Python due to its versatility and extensive documentation.

## **Quick Start**

```python
# import the necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
```

NumPy (`numpy`) is imported because Matplotlib relies on it for efficient numerical computations and handling arrays, which are often used as input data for plotting. Matplotlib's `pyplot` module (`matplotlib.pyplot`) provides a convenient, MATLAB-like interface for creating and customizing plots. The core `matplotlib` library (`matplotlib`) is imported to access advanced configurations and features, such as setting global styles or accessing lower-level plotting components. These libraries together enable seamless data manipulation and visualization in Python.

```python
X = np.linspace(0, 2*np.pi, 100)
Y = np.cos(X)
fig, ax = plt.subplots()
ax.plot(X, Y, color='green')
fig.savefig("figure.pdf")
plt.show()
```

## **Anatomy of Figures**

In Matplotlib, the **anatomy of a figure** refers to the hierarchical structure of components that make up a plot, providing a framework for creating and customizing visualizations. The **Figure** is the top-level container that holds all elements, acting as a canvas for one or more **Axes**, which are individual plotting areas where data is visualized. Each **Axes** contains **Axis** objects (x and y axes) that manage ticks, tick labels, and axis labels, defining the scale and boundaries of the plot. Additional components like the **Title** (describing the plot), **Legend** (identifying data series), and **Artists** (visual elements like lines, text, or patches) enhance clarity and aesthetics. This modular structure allows precise control over plot customization through Matplotlib’s object-oriented API, enabling complex layouts with multiple subplots. Understanding this anatomy is essential for effectively manipulating plot elements to meet specific visualization needs.

## **Subplot Layout**

In Matplotlib, a subplot layout is a grid-based arrangement of multiple **Axes** (plotting areas) within a single **Figure**, allowing the creation of complex, multi-panel visualizations. The `plt.subplot()` or `plt.subplots()` functions are used to define the layout by specifying the number of rows and columns in the grid, with each cell containing an individual plot. For example, `plt.subplots(2, 3)` creates a 2x3 grid of six subplots, each accessible for customization. Subplot layouts are ideal for comparing datasets or displaying related visualizations side by side. Matplotlib provides flexibility to adjust spacing, share axes, and customize each subplot independently to suit specific needs.

```python
# the subplotting using matplotlib using eight different variables
fig, axs = plt.subplots(2, 4)
axs[0, 0].plot(X, Y)
axs[0, 1].plot(X, Y)
axs[0, 2].plot(X, Y)
axs[0, 3].plot(X, Y)
axs[1, 0].plot(X, Y)
axs[1, 1].plot(X, Y)
axs[1, 2].plot(X, Y)
axs[1, 3].plot(X, Y)
plt.show()
```

## **Basics Plots**

```python
# 1. Line plots
line_plot= np.random.randn(100)
plt.plot(line_plot)
plt.show()
```

```python
# 2. Scatter plots
x= np.random.randn(100)
y= np.random.randn(100)
plt.scatter(x, y)
plt.show()
```

```python
# 3. Barplots
x= np.arange(10)
y= np.random.randn(10)
plt.bar(x, y)
plt.show()
```

```python
# 4. imshow plots
x= np.random.randn(100, 100)
plt.imshow(x)
plt.show()
```

```python
# 5. countour plots
x= np.random.randn(100, 100)
plt.contour(x)
plt.show()
```

```python
# 6. pcolormech plots
x= np.random.randn(100, 100)
plt.pcolormesh(x)
plt.show()
```

```python
# 7.pie charts
y = np.array([35, 25, 25, 15])
plt.pie(y)
plt.show()
```

```python
# 9. Text in plots
x = np.arange(-10, 10, 0.01)
y = x**2
# adding text inside the plot
plt.text(-5, 60, 'Parabola $Y = x^2$', fontsize = 22)
plt.plot(x, y, c='g')
plt.xlabel("X-axis", fontsize = 15)
plt.ylabel("Y-axis",fontsize = 15)
plt.show()
```

```python
# 10.fill charts 
x = np.array([1, 2, 3, 4])
y = x*2
plt.plot(x, y)
x1 = [2, 4, 6, 8]
y1 = [3, 5, 7, 9]
plt.plot(x, y1, '-.')
plt.xlabel("X-axis data")
plt.ylabel("Y-axis data")
plt.title('multiple plots')
plt.fill_between(x, y, y1, color='green', alpha=0.5)
plt.show()
```
