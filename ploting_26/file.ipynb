```python
# Import necessary libraries
import ipywidgets as widgets
from IPython.display import display, HTML
import matplotlib.pyplot as plt
import numpy as np

# Sample code cell
x = np.linspace(0, 2*np.pi, 100) 
y = np.sin(x)

# Simple plot
plt.plot(x, y)
plt.title('Simple Sine Wave')

# Animated plot
@widgets.interact
def plot_wave(amplitude=widgets.FloatSlider(min=0, max=5, step=0.1, value=1)):
  plt.plot(x, amplitude*y)
  plt.title(f'Animated Sine Wave of Amplitude {amplitude}')
  display(plt.gcf())

# Interactive plot
slider = widgets.FloatSlider(min=0, max=2*np.pi, step=0.1, value=np.pi/2)
@widgets.interact
def plot_wave(xval=slider):
  plt.plot(x, y)
  plt.plot([xval, xval], [-1, 1], '--r')
  plt.title(f'Interactive plot at x={xval}')
  display(plt.gcf())

# Convert notebook to HTML
from nbconvert import HTMLExporter
exporter = HTMLExporter()
exporter.template_file = 'basic'
(body, resources) = exporter.from_filename('./notebook.ipynb')
with open('./notebook.html', 'w') as f:
  f.write(body)
```
