## **Assingnment**

### **What is the difference between int64 ,float32  64 and 16 ..... In ka maltalb kia hy?**

Acha, jab hum `int64`, `float32`, `float64`, aur `float16` ki baat karte hain, to ye **data types** hain jo numbers ko store karne ka tareeqa batate hain. Ye numbers store karne ki **size (bits)** aur **precision (accuracy)** define karte hain.

### **1. int64 (Integer 64-bit)**
- **Int** ka matlab hai **integer** (matlab poore numbers bina decimal ke, jaise 1, 2, -5, 100).
- `int64` ka matlab hai **64-bit integer**, jo bohot bare numbers store kar sakta hai.
- Iska range: \(-2^{63}\) se \(2^{63}-1\) tak hota hai.

#### **Example in Python**
```python
import numpy as np

x = np.int64(1000000000000)  # Ek bara number store karna
print(x, type(x))  # Output: 1000000000000 <class 'numpy.int64'>
```

---

### **2. float32 (Floating Point 32-bit)**
- **Float** ka matlab hai **decimal numbers** (jaise 3.14, -2.5, 0.0001).
- `float32` ek **32-bit floating point number** hai, jo decimals ko store karta hai.
- Chhoti memory use karta hai, lekin precision thodi kam hoti hai.

#### **Example in Python**
```python
y = np.float32(3.141592653589793)
print(y, type(y))  # Output: 3.1415927 <class 'numpy.float32'>
```
*Note:* `float32` sirf **7 decimal places** tak accuracy rakhta hai.

---

### **3. float64 (Floating Point 64-bit)**
- `float64` ek **64-bit floating point number** hai.
- Ye `float32` se zyada accurate hota hai aur **15-16 decimal places** tak store kar sakta hai.

#### **Example in Python**
```python
z = np.float64(3.141592653589793)
print(z, type(z))  # Output: 3.141592653589793 <class 'numpy.float64'>
```
*Note:* `float64` ki precision `float32` se zyada hoti hai.

---

### **4. float16 (Floating Point 16-bit)**
- `float16` ek **16-bit floating point number** hai.
- Chhoti memory use karta hai, lekin precision sirf **3-4 decimal places** tak hoti hai.
- Mostly **machine learning** aur **image processing** me use hota hai jahan memory save karni hoti hai.

#### **Example in Python**
```python
a = np.float16(3.141592653589793)
print(a, type(a))  # Output: 3.14 <class 'numpy.float16'>
```
*Note:* `float16` me sirf **3-4 decimal places** tak precision hoti hai.

---

### **Comparison Table**
| Data Type | Bits | Number Type | Precision |
|-----------|------|-------------|------------|
| int64     | 64   | Integer      | Poore numbers, bada range |
| float32   | 32   | Floating Point | ~7 decimal places |
| float64   | 64   | Floating Point | ~15-16 decimal places |
| float16   | 16   | Floating Point | ~3-4 decimal places |
