# **What is an array ?**


An array is a data structure that stores a collection of elements, each identified by an index. The elements in an array are usually of the same type, and the array has a fixed size, determined when the array is created. In PowerShell, arrays are created by placing the elements inside parentheses and separated by commas. The elements can be of any type, including other arrays, creating a multi-dimensional array. Arrays can be manipulated using various methods and properties, such as .Length, .Where(), and .ForEach().

```python
a=np.arange(10)         # its is one dimensional array
a
```

```python
a.shape
```

```python
a_1=a[np.newaxis, :]                        # this is a two dimensional array
a_1
```

```python
a_1.shape
```

```python
a_2=a[:, np.newaxis]                        # this is a two dimensional array
a_2
```

```python
a_3=a_2[np.newaxis, :]                      # this is a three dimensional array
a_3

a_3.shape
```

## **Why do we have 1D,2D and 3D arrays and where we need them in data sicence ?**

### **1D Arrays**
1D Arrays are used when we need to store a single row or column of data. They are useful when we need to perform operations on a single row or column of data. For example,we can use 1D arrays to store the scores of a student in a subject. We can use the array to perform operations like finding the maximum score, minimum score, average score, etc.

### **2D Arrays (Matrices)**

2D Arrays are used when we need to store data in a table format.They are useful when we need to perform operations on multiple rows and columns of data. For example,we can use 2D arrays to store the scores of students in a subject. We can use the array to perform operations like finding the maximum score, minimum score, average score, etc.

### **3D Arrays (Cubes)**

3D Arrays are used when we need to store data in a 3D format.They are useful when we need to perform operations on multiple rows, columns, and layers of data.For example, we can use 3D arrays to store the scores of students in multiple subjects.We can use the array to perform operations like finding the maximum score, minimum score, average score,etc.
