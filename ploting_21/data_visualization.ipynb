# Data Visualization

```python
# now import the necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# now import the dataset of the titanic form the seaborn library
df = sns.load_dataset('titanic')
df.head()
```

```python
df.plot(kind='scatter', x='age', y='fare',color='red')
```

```python
# now we change the color using hex code
df.plot(kind='scatter', x='age', y='fare',color='#ff0000')
```

```python
df.plot(kind='scatter', x='age', y='fare',color='#1b4242')
```

To select the hex color form the google or online we search for the color picker and color picker google extension to select the requied color hex and use it accoring to over requirment.

```python
df.groupby('sex')['fare'].mean().plot(kind='bar',color='#1b4242')
plt.xlabel('Gender')
plt.ylabel('Fare')
```

```python
# now we create the plot on the seaborn library
sns.barplot(x='sex', y='fare', data=df, hue='class', palette='Set2')
```

```python
sns.barplot(x='sex', y='age', data=df, hue='class', palette='Set2')
```

```python
sns.boxplot(x='sex', y='age', data=df, hue='class', palette='Set2')
```

```python
# now we create the plot to check the relationship between age and fare and see which age group has pay the high fair  to the 
# journy
sns.scatterplot(x='age', y='fare', data=df, hue='class', palette='Set2')
```

```python
# now we see that the pessanger  who  has pay the highest is male or female
sns.scatterplot(x='age', y='fare', data=df, hue='sex', palette='Set2')
```

```python
# now we have to see  that the passanger with the highest age in the titanic journy is male or female
sns.scatterplot(x='age', y='fare', data=df, hue='sex',color='#ff0000')
```
