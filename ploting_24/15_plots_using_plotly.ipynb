# 15_plots in Plotly

```python
# import the necessary libraries 
import plotly.express as px
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
# now we load the dataset for plotting
df=sns.load_dataset('iris')
df.head()
```

```python
df.info()
```

# 1. Scatter plot

```python
fig_1=px.scatter(df,x='sepal_length',y='sepal_width',color='species')
fig_1.show()
```

# 2. Line Plot

```python
fig_2=px.line(df,x='sepal_length',y='sepal_width',color='species')
fig_2.show()
```

```python
# as we see above the graph does not make any sense so we simplt sort the values 
df_line=df.sort_values('sepal_length')
fig_2=px.line(df_line,x='sepal_length',y='sepal_width',color='species')
fig_2.show()
```

#  3. Barplot

```python
fig_3=px.bar(df,x='species',y='sepal_width')
fig_3.show()
```

```python
# in the above we see we have bars divided into pieces based on the species but we know that the bars shows on average 
# values of  the species
df_bar=df.groupby('species').mean().reset_index()
fig_3=px.bar(df_bar,x='species',y='sepal_width')
fig_3.show()
```

```python
df_bar  # here we see that we have average values of each species (length and widths)
```

# 4. Boxplot

```python
fig_4=px.box(df,x='species',y='sepal_width',color='species')
fig_4.show()
```

# 5. Violin Plot

```python
fig_5=px.violin(df,x='species',y='sepal_width',color='species')
fig_5.show()
```

```python
# you can also show the boxplot in the violin plot by using the below code
fig_5_1= px.violin(df,x='species',y='sepal_width',color='species',box=True)
fig_5_1.show()
```

# 6. Histgram

```python
fig_6=px.histogram(df,x='sepal_width')
fig_6.show()
```

```python
fig_6=px.histogram(df,x='sepal_width',color='species')
fig_6.show()
```

# 8. 3-D scatter plot

```python
fig_8=px.scatter_3d(df,x='sepal_length',y='sepal_width',z='petal_width',color='species')
fig_8.show()
```

# 9. Area Chart

```python
fig_9=px.area(df,x='sepal_length',y='sepal_width',color='species')
fig_9.show()
```

# 10. Bubble Chart

```python
fig_10=px.scatter(df,x='sepal_length',y='sepal_width',color='species',size='petal_width')
fig_10.show()
```

# 11. Sunburn Chart

```python
df_sunburst = df.groupby(['species', 'petal_width']).size().reset_index(name='counts')
fig_11 = px.sunburst(df_sunburst, path=['species', 'petal_width'], values='counts', color='petal_width')
fig_11.show()
```

# 12. parallal cordinate plot

```python
import plotly.express as px
import pandas as pd
```

```python
# Load sample dataset
df = px.data.iris()

# Convert categorical column 'species' to numerical values
df['species_id'] = df['species'].astype('category').cat.codes

# Create Parallel Coordinates Plot
fig = px.parallel_coordinates(df, 
                              dimensions=['sepal_width', 'sepal_length', 'petal_width', 'petal_length'], 
                              color='species_id',  # Now a numerical column
                              color_continuous_scale=px.colors.sequential.Inferno)

# Show the plot
fig.show()
```

```python
fig_12=px.parallel_coordinates(df,color='species')
fig_12.show()
```

# 13. Density contour plot

```python
fig_13=px.density_contour(df,x='sepal_length',y='sepal_width',color='species')
fig_13.show()
```

# 14. Ternary plot

```python
fig_14=px.scatter_ternary(df,a='sepal_length',b='sepal_width',c='petal_width',color='species')
fig_14.show()
```
