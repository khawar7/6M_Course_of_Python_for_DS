# Seaborn and Plotting

```python
# import the necessary liberaries we need for this 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

# 1. Load the dataset 

```python
df=sns.load_dataset('titanic')
```

```python
# check the data set
df.info ()
```

# 2. Type of plot

```python
# scatter plot
sns.scatterplot(x='age',y='fare',data=df)
```

# 3. X labels

```python
sns.scatterplot(x='age',y='fare',data=df)
# define the x labels as follows
plt.xlabel('Age')
```

# 4. Y labels

```python
sns.scatterplot(x='age',y='fare',data=df)
# define the x labels as follows
plt.xlabel('Age')
# define the y labels as follows
plt.ylabel('Fare')
```

# 5. Grouping (color/hue)

```python
sns.scatterplot(x='age',y='fare',data=df,hue='sex')
# define the x labels as follows
plt.xlabel('Age')
# define the y labels as follows
plt.ylabel('Fare')
```

```python
# now we do the grouping using the color parameter
sns.scatterplot(x='age',y='fare',data=df,hue='class')
# define the x labels as follows
plt.xlabel('Age (years)')
# define the y labels as follows
plt.ylabel('Fare (Pounds)')
```

# 6. Size of plot

```python
fig=sns.scatterplot(x='age',y='fare',data=df,hue='class')
# define the x labels as follows
plt.xlabel('Age (years)')
# define the y labels as follows
plt.ylabel('Fare (Pounds)')
# how the size of the plot
fig.sizes=(10,6)
```

### For A4 size paper we have the fig.size(8,6).

# 7. Color Pallat 

```python
# now i have to change the color of the scatter point in the plot for this i have define the color palette
palette={'First':'red','Second':'green','Third':'blue'}
fig=sns.scatterplot(x='age',y='fare',data=df,hue='class',palette=palette)
# define the x labels as follows
plt.xlabel('Age (years)')
# define the y labels as follows
plt.ylabel('Fare (Pounds)')
# how the size of the plot
fig.sizes=(10,6)
# now i have to add the title of the plot
plt.title('Titanic Passangers fare by class of different ages')
# now i am going to save the plot 
plt.savefig('Titanic Passangers fare by class of different ages.png',dpi=600)
# now save the file in svg format
plt.savefig('Titanic Passangers fare by class of different ages.svg',dpi=600)
```

# 8. Sytle of Plot

```python
# now i have to change the color of the scatter point in the plot for this i have define the color palette
palette={'First':'red','Second':'green','Third':'blue'}
fig=sns.scatterplot(x='age',y='fare',data=df,hue='class',palette=palette)
# define the x labels as follows
plt.xlabel('Age (years)')
# define the y labels as follows
plt.ylabel('Fare (Pounds)')
# how the size of the plot
fig.sizes=(10,6)
# now i have to add the title of the plot
plt.title('Titanic Passangers fare by class of different ages')
# now i am going to save the plot 
plt.savefig('Titanic Passangers fare by class of different ages.png',dpi=600)
# now save the file in svg format
plt.savefig('Titanic Passangers fare by class of different ages.svg',dpi=600)
# now i am going to add the theme of change the style of the plot by changing the style, font_scale and font
sns.set_theme(style="darkgrid",font_scale=1.0,font="serif")
```
