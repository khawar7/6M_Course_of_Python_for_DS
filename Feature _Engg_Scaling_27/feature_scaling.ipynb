# 1. Min-Max Scaling

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# now we create over own sample data 
df = pd.DataFrame({'col_A': [1, 2, 3, 4, 5], 'col_B': [10, 20, 30, 40, 50]})
df
```

```python
# now the min-max scaling is implemented
scaler=MinMaxScaler()
df['scaled_A']=scaler.fit_transform(df[['col_A']])
df
```

```python
# now the min-max scaling is implemented
scaler=MinMaxScaler()
df['scaled_B']=scaler.fit_transform(df[['col_B']])
df
```

# 2. Standard Scaler or Z-Score normalization

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# now we create over own sample data
data_1=pd.DataFrame({'col_A': [10, 20, 30, 40, 50], 'col_B': [100, 200, 300, 400, 500]})
data_1
```

```python
# now the standard scaling is implemented
standard_scaler=StandardScaler()
data_1['scaled_A']=standard_scaler.fit_transform(data_1[['col_A']])
data_1
```

```python
# now the standard scaling is implemented
standard_scaler=StandardScaler()
data_1['scaled_B']=standard_scaler.fit_transform(data_1[['col_B']])
data_1
```

# 3. Robust Scaler

```python
import pandas as pd
from sklearn.preprocessing import RobustScaler

# now we create over own sample data
data_1=pd.DataFrame({'col_A': [10, 20, 30, 40, 50], 'col_B': [100, 200, 300, 400, 500]})
data_1
```

```python
# now the robust scaling is implemented
robust_scaler=RobustScaler()
data_1['scaled_A']=robust_scaler.fit_transform(data_1[['col_A']])
data_1
```

```python
# now the standard scaling is implemented
robust_scaler=RobustScaler()
data_1['scaled_B']=robust_scaler.fit_transform(data_1[['col_B']])
data_1
```

# 4. Lograthemic scaling

```python
import numpy as np
import pandas as pd

# now we take the radom data with the outliers
data_2=pd.DataFrame({'col_A': [10, 20, 30, 40, 50]})
data_2
```

```python
# now the log scaling is implemented
data_2['scaled_A']=np.log(data_2['col_A'])
data_2
```

```python
data_2['scaled_A_2']=np.log2(data_2['col_A'])
data_2
```

```python
data_2['scaled_A_10']=np.log10(data_2['col_A'])
data_2
```

```python
# now create the boxplot of the data_2 and see the outliers in it 
import seaborn as sns
sns.boxplot(data_2,x='scaled_A_10')
```
