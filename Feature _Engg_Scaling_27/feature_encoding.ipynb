# Feature Encoding

# 1. One hot encoding

```python
# import the necessary libary
import pandas as pd
```

```python
# now we make the sample data of over own
df=pd.DataFrame({
    'gender':['male','female','female','male'],
    'status':['single','married','single','married']
})
df
```

```python
# now do the one hot encoding on the df dataset
encoded_df=pd.get_dummies(df)
encoded_df
```

# 2. Label encoding

```python
# imprort the necessary libary for the label encoding
import sklearn.preprocessing as label_encoder
```

```python
# take the sample data of over own
data_1=pd.DataFrame({
    'gender':['male','female','female','male'],
    'status':['single','married','single','married']
})
data_1
```

```python
# now we do the label encoding on the data_1
le = label_encoder.LabelEncoder()
data_1['gender']=le.fit_transform(data_1['gender'])
data_1
```

```python
# now we do the label encoding on the data_1
le = label_encoder.LabelEncoder()
data_1['status']=le.fit_transform(data_1['status'])
data_1
```

# 3. Ordinal encoding

```python
# imprort the necessary libary for the label encoding
import sklearn.preprocessing as ordinal_encoder
```

```python
# take the sample data of over own
data_2=pd.DataFrame({
    'gender':['male','female','female','male'],
    'status':['single','married','single','married']
})
data_2
```

```python
# now we do the ordinal encoding on the data_1
oe=ordinal_encoder.OrdinalEncoder()
data_2['gender']=oe.fit_transform(data_2[['gender']])
data_2
```

```python
# now we do the ordinal encoding on the data_1
oe=ordinal_encoder.OrdinalEncoder()
data_2['status']=oe.fit_transform(data_2[['status']])
data_2
```
