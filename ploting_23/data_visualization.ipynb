# Boxenplot

```python
# now i convert the all of the above as i do it for the scatter plot into the boxplot
plt.figure(figsize=(12,6))
parameter={'First':'red','Second':'blue','Third':'green'}
sns.set_theme(style="darkgrid",font_scale=1.1,font='Times New Roman')
# now i want to change the color of  the scatter points in the scatter points  using the parameter vector
sns.boxenplot(x="sex", y="age",hue='class', data=df,palette=parameter)
# now we define the labels of the plot using the xlabel and ylabel 
plt.xlabel('sex')
plt.ylabel('age')
# now we define the title of the plot
plt.title('Relationship between Age and sex')
# now we simple foucs on saving the plot and keeping in mind the resolution of the plot we want to save it into the pnj file 
# or any other format
plt.savefig('Age and Fare scatter_plot.png',dpi=600)
# change the legend postion and the title
plt.legend(loc='upper right',title='Class')
# now we simplt change the x ticks labels and rename them 
plt.xticks(ticks=[0,1],labels=['Male','Female'])
plt.xticks(rotation=45)
```

# Barplot

```python
# now i convert the all of the above as i do it for the scatter plot into the boxplot
plt.figure(figsize=(12,6))
parameter={'First':'red','Second':'blue','Third':'green'}
sns.set_theme(style="darkgrid",font_scale=1.1,font='Times New Roman')
# now i want to change the color of  the scatter points in the scatter points  using the parameter vector
sns.barplot(x="sex", y="age",hue='class', data=df,palette=parameter,capsize=0.2)        # how the capsize work in the barplot as error bars
# now we define the labels of the plot using the xlabel and ylabel 
plt.xlabel('sex')
plt.ylabel('age')
# now we define the title of the plot
plt.title('Relationship between Age and sex')
# now we simple foucs on saving the plot and keeping in mind the resolution of the plot we want to save it into the pnj file 
# or any other format
plt.savefig('Age and Fare scatter_plot.png',dpi=600)
# change the legend postion and the title
plt.legend(loc='upper right',title='Class')
# now we simplt change the x ticks labels and rename them 
plt.xticks(ticks=[0,1],labels=['Male','Female'])
plt.xticks(rotation=45)
```

# Violin plot

```python
# now i convert the all of the above as i do it for the scatter plot into the boxplot
plt.figure(figsize=(10,6))
parameter={'First':'red','Second':'blue','Third':'green'}
sns.set_theme(style="darkgrid",font_scale=1.1,font='Times New Roman')
# now i want to change the color of  the scatter points in the scatter points  using the parameter vector
sns.violinplot(x="sex", y="age",hue='class', data=df,palette=parameter)        
# now we define the labels of the plot using the xlabel and ylabel 
plt.xlabel('sex')
plt.ylabel('age')
# now we define the title of the plot
plt.title('Relationship between Age and sex')
# now we simple foucs on saving the plot and keeping in mind the resolution of the plot we want to save it into the pnj file 
# or any other format
plt.savefig('Age and Fare scatter_plot.png',dpi=600)
# change the legend postion and the title
plt.legend(loc='upper right',title='Class')
# now we simplt change the x ticks labels and rename them 
plt.xticks(ticks=[0,1],labels=['Male','Female'])
plt.xticks(rotation=45)
```
