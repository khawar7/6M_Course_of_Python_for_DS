# **Test of Homogenity**

## **What is homogenity, and how is it related to the concept of a "homogeneous" distribution?**
Homogeneity is a property of a distribution that means that the distribution is unchanged under a translation of the coordinate system. In other words, if we shift the coordinate system by a certain amount, the distribution remains the same. This is a fundamental concept in probability theory and statistics, and it is related to the concept of a "homogeneous" distribution, which is a distribution that is invariant under translations of the coordinate system.

```python
# import the necessary libraries
import numpy as np 
import pandas as pd
from scipy import stats
```

### **Levene's Test**

```python
# Sample data
group1 = [2.3, 3.4, 4.5, 2.3, 3.4]
group2 = [1.2, 2.2, 3.2, 2.2, 2.3]

# Perform Levene's Test
w_stat, p_val = stats.levene(group1, group2)

print("w-statistic:", w_stat)
print("p-value:", p_val)

# print the results based on if else conditions
if p_val > 0.05:
    print(f'p-value: {p_val}, Variances are equal (fail to reject H0)')
else:
    print(f'p-value: {p_val}, Variances are not equal (reject H0)')
```

### **Bartlet Test**

```python
# Sample data
group1 = [2.3, 3.4, 4.5, 2.3, 3.4]
group2 = [1.2, 2.2, 3.2, 2.2, 2.3]

# Perform Bartlett's Test
t_stat, p_val = stats.bartlett(group1, group2)

print("t-statistic:", t_stat)
print("p-value:", p_val)

# print the results using if else conditions
if p_val > 0.05:
    print(f'p-value: {p_val}, Variances are equal (fail to reject H0)')
else:
    print(f'p-value: {p_val}, Variances are not equal (reject H0)')
```
