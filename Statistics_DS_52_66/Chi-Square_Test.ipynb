# **Chi-Square test**

## **Why is chi-square test used for?**
The chi-square test is used to determine whether there is a significant association between two categorical variables. It
is used to test the null hypothesis that there is no association between the variables.

```python
# import the neessary libraries
import pandas as pd
import numpy as np
import seaborn as sns
from scipy import stats
```

```python
# load the titanic built-in dataset
titanic = sns.load_dataset('titanic')
titanic.head()
```

### **Define the Hypothesis of Chi-Square Test.**
**Null Hypothesis (H0**): There is no association between the gender(`sex`) & survial (`survived`) on the Titanic.

**Alternative Hypothesis (H1)**: There is a association between the gender(`sex`) & survial (`survived`) on the Titanic.

```python
# Perform the chi-square test

# Contingency table
contingency_table = pd.crosstab(titanic['survived'], titanic['sex'])
contingency_table
```

```python
# Chi-square test
chi2, p, dof, expected = stats.chi2_contingency(contingency_table)
print(f'Chi-square value: {chi2}')
print(f'p-value: {p}')
print(f'Degrees of freedom: {dof}')
print(f'Expected frequencies: {expected}')
```

#### We reject the null hypothesis, because the p-value (1.1973570627755645e-58) is very less than 0.05.
