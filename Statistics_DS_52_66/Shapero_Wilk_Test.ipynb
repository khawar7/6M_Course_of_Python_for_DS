# **Normal Distribution Test**

### What is the test to check the normality of the data?
A. Kolmogorov-Smirnov test\
B. Shapiro-Wilk test\
C. Anderson-Darling test

```python
# load the necessary libraries
import numpy as np
import pandas as pd
import seaborn as sns
from scipy import stats
```

```python
# load the titanic built-in dataset
df=sns.load_dataset('titanic')
df.head()
```

```python
# Check the data is normal or not by plotting
sns.distplot(df['fare'], hist=True, kde=True,
             bins=int(180/5), color = 'darkblue',
             hist_kws={'edgecolor':'black'},
             kde_kws={'linewidth': 4})
```

### Define the Hypothesis
**Null Hypothesis**: H0 = μ = 0 (The population mean is equal to 0)\
**Alternative Hypothesis**: H1 = μ ≠ 0 (The population mean is not equal to 0)

```python
# Now check by using the Shapiro-Wilk test
stat, p = stats.shapiro(df['fare'])
print('Statistics=%.3f, p=%.3f' % (stat, p))

# interpret
if p > 0.05:
    print('Data is normally distributed')
else:
    print('Data is not normally distributed')
```

```python
# now check the normality using the kolmogorov-smirnov test
stat, p = stats.kstest(df['fare'], 'norm')
print('Statistics=%.3f, p=%.3f' % (stat, p))

# interpret
if p > 0.05:
    print('Data is normally distributed')
else:
    print('Data is not normally distributed')
```
