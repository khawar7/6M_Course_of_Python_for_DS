# **One Way ANOVA**

### **What is ANOVA?**
ANOVA (Analysis of Variance) is a statistical technique used to compare means of three or more groups to find out if at least one group mean is different. It is used to determine if there is a significant difference between the means of three or more groups.

### **What is the purpose of ANOVA?**
The purpose of ANOVA is to determine if there is a significant difference between the means of three or more groups. It is used to test the null hypothesis that all group means are equal,and to determine if the observed differences between the means are due to chance or if they are statistically significant

```python
# import the necessary libraries
import numpy as np
from scipy import stats
import pandas as pd
```

## **ANOVA (Analysis of Variance)**

```python
# Sample data

a = [2.3, 3.4, 4.5, 2.3, 3.4]
b = [11.2, 12.2, 13.2, 12.2, 12.3]
c = [22.3, 23.4, 24.5, 22.3, 23.4]

# perform one-way ANOVA
f_stat, p_val = stats.f_oneway(a, b, c)

# print the results base on if else conditions
if p_val > 0.05:
    print(f'p-value: {p_val}, Means are equal (fail to reject H0)')
else:
    print(f'p-value: {p_val}, Means are not equal (reject H0)')
```
