# **Linear Transformation in Python**

```python
# import the necessary libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# create a vector
v = np.array([3,3])
v
```

```python
# Define the rotation matrix for 45 degree rotation
theta = np.radians(55)  # Converting 45 degrees to radians
rotation_matrix = np.array([[np.cos(theta), -np.sin(theta)], 
                            [np.sin(theta), np.cos(theta)]])

print(rotation_matrix)
print("------------------------")
print(v)
# Applying the transformation
v_transformed = np.dot(rotation_matrix, v)
v_transformed
```

```python
# Plotting the original and transformed vectors
plt.figure()
plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='r', label='Original Vector')
plt.quiver(0, 0, v_transformed[0], v_transformed[1], angles='xy', scale_units='xy', scale=1, color='b', 
           label='Transformed Vector')
plt.xlim(-1, 5)
plt.ylim(-1, 5)
plt.grid()
plt.title('2D Vector Transformation')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.legend()
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```
