# **Vectors in Python**

```python
# import the necessary library
import numpy as np
import matplotlib.pyplot as plt
```

```python
# create a vector
a = np.array([2, 3])
print(a)
```

```python
plt.figure()
plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='g')
plt.xlim(0, 5)
plt.ylim(0, 5)
plt.grid()
plt.title('Vector Plot with Arrowhead')
plt.xlabel('x')
plt.ylabel('y')
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```
