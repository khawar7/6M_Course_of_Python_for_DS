# **SVD in Python**

```python
# import the necessary libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# generate a matrix  of 4x3 dataset
A = np.random.rand(4,3)
A
```

```python
# lets apply SVD here
U, Sigma, VT = np.linalg.svd(A)
# print each matrix separately
print("U Matrix=", U)
print("--------------------------------")
print("Sigma Matrix=",Sigma)
print("--------------------------------")
print("VT Matrix=",VT)
```
