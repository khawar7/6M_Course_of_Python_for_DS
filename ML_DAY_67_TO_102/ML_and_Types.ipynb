# **Machine Learning and Types of ML**

## **What is Machine Learning?**

A machine that can learn from the data given to it and make predictions or decision based on the data that is not explicitly programmed to do so.

For example, let we have a data structure of pattern that is:

2 x 1 = 2

2 x 2 = 4

2 x 3 = 6

2 x 4 = ?

For the above pattern the machine learn from the structural data the pattern of the data and then make the prediction of it.
So the answer is 8.

## **Steps followed by the Machine Learning Algorithm**

Data ----> Learning ----> Prediction ----> Decision

first of all is the data given to the machine, then the machine learn the patterns form the data and then make the prediction based on the data and finally make the decision made by you.

## **Types of Machine Learning**
---------------------------------------------------------------------------------------------------------------------------------
1. **Supervised Learning**

when a machine is trained on a labeled data set, to predict the outcome of unseen data based on the given dataset. This is similar to the way a student is taught by a teacher by giving the answer to the questions or problems and than  the predict the answer to the unseen questions.

labeled data ----> training the data by giving the answers to the questions ----> predict the answer to the unseen questions ----> the make a decision or create a report.

Types o supervised learning data are:
- classification (for categories)
- regression (for numerical data)

Types of supervised learning algorithms are:
- logistics regression
- kernal svm
- k-nearest neighbors
- support vector machine
- naive bayes
- decision tree classification
- random forest classification
---------------------------------------------------------------------------------------------------------------------------------

2. **Unsupervised learning / clustering**

When a machine is trained on an unlabeled data set, to find the hidden patterns in the input data without any guidance. Similar to the way a student is given a problem and asked to find the solution without any help form the teacher.

unlabeled data ----> training the data without any answers and the machine make clusters of the data ----> Predict the hidden patterns in the unseen data ----> make a decision or create a report.

Types of unsupervised learning algorithms are:
- k means clustering
- hierarchical clustering
- Probabilistic clustering
---------------------------------------------------------------------------------------------------------------------------------

3. **Semi-supervised learning**

A hybrd approach that combines both supervised and unsupervised learning techniques. In this approach, the machine is trained on a small amount of labeled data and a large amount of unlabeled data. The labeled data helps the machine to learn the patterns and the unlabeled data helps to improve the accuracy of the model.

labeled + unlabeled data ----> training the data with some answers and some without answers ----> predict the outcome of unseen data ----> make a decision or create a report.

----------------------------------------------------------------------------------------------------------------------------------

4. **Reinforcement learning**

A technique where the machine learns to make decisions by interactin with an enviroment or by trail or error method, and receiving feedback in the form of rewards or penalties. The machine learns to take actions that maximize the cumulative reward over time.

input data + enviroment ----> training the data by taking actions and receiving feedback / trail or error method ----> predict the best action to take in unseen situations ----> make a decision or create a report.

Types of reinforcement learning algorithms are:
1. model free learning methods
- Q-learning
- Policy Optimization
2. model based learning methods
- Learn the model
- Given the Model
----------------------------------------------------------------------------------------------------------------------------------
