# **Machine Learning and Types of ML**

## **What is Machine Learning?**

A computer program to whcih we feed the data that is the training data and the program learn form it by finding the patterns and trends in it, and then make the prediction based on the new data that is given to it.

For example, let we have a data structure of pattern that is:

2 x 1 = 2

2 x 2 = 4

2 x 3 = 6

2 x 4 = ?

For the above pattern the machine learn from the structural data the pattern of the data and then make the prediction of it.
So the answer is 8.

## **Steps followed by the Machine Learning Algorithm**

Computer program---------> Data(trainging data)-------> Learning(patterns and trends)-------> Prediction/Decision(new testing data).

## **Types of Machine Learning**
-------------------------------------------------------------------------------------------------------------------------------
1. **Supervised Learning**

A Computer program or a  machine program is trained on a labeled data set or training data, to predict the outcome of unseen data or test dataset based on the given dataset. For example a student is taught by a teacher by giving the answer to the questions or problems and than  the predict the answer to the unseen questions.

Types o supervised learning data are:
- classification (for categories)
- regression (for numerical data)

Types of supervised learning algorithms are:
- logistics regression
- kernal svm
- k-nearest neighbors
- support vector machine
- naive bayes
- decision tree classification
- random forest classification
-------------------------------------------------------------------------------------------------------------------------------
2. **Unsupervised learning / clustering**

When a computer program or machine program  is trained on an unlabeled data set, to find the hidden patterns in the input data without any guidance.For example a student is given a problem and asked to find the solution without any help form the teacher.

Types of unsupervised learning algorithms are:
- k means clustering
- hierarchical clustering
- Probabilistic clustering
-------------------------------------------------------------------------------------------------------------------------------
3. **Semi-supervised learning**

A hybrd approach that combines both supervised and unsupervised learning techniques. In this approach, the machine is trained on a small amount of labeled data and a large amount of unlabeled data. The labeled data helps the machine to learn the patterns and the unlabeled data helps to improve the accuracy of the model.

-------------------------------------------------------------------------------------------------------------------------------
4. **Reinforcement learning**

A technique where the machine learns to make decisions by interaction with an enviroment or by trail or error method, and receiving feedback in the form of rewards or penalties. The machine learns to take actions that maximize the cumulative reward over time.

Types of reinforcement learning algorithms are:
1. model free learning methods
- Q-learning
- Policy Optimization
2. model based learning methods
- Learn the model
- Given the Model
-------------------------------------------------------------------------------------------------------------------------------
