# **Training and Testion of Data in Machine Learning Model?**

**Training and Testing Data in Machine Learning**

**Definition:**  
- `Training data` is the portion of the dataset used to teach or train the machine learning model by allowing it to learn patterns, relationships, and features from the data.  
- `Testing data` is a separate, unseen portion of the dataset used to evaluate the performance and generalization ability of the trained model on new data.

**Important points:**

- **Purpose:**  
  - Training data is used to build and fit the model.  
  - Testing data is used to assess how well the model performs on unseen data and to estimate its real-world effectiveness.

- **Exposure:**  
  - The model learns from the training data and adjusts its parameters accordingly.  
  - The testing data is kept hidden from the model during training to provide an unbiased evaluation.

- **Data distribution:**  
  - Both training and testing data should ideally come from the same distribution to ensure meaningful evaluation.  
  - Testing data simulates new, real-world data that the model will encounter after deployment.

- **Size:**  
  - Training data is usually larger (commonly around 70-80% of the total dataset) to provide sufficient examples for learning.  
  - Testing data is smaller (commonly 20-30%) but must be large enough to yield statistically significant evaluation results.

- `**Features**`: The individual measurable properties or characteristics of the data that serve as input variables for the model to make predictions. For example we have to predict the price of the house in the ml model are we are give the features like size of plot, house consturction size, material used, loction of the house and many more. More the features better the model performance.

- `**Model**`: A trained program that uses an algorithm to identify patterns in data and make predictions or classifications on new, unseen data.

# **Overfitting and Underfitting of Machine Learning Models**

**`Overfitting`** a common problem in machine learning where a model learns the training data too well, including its noise and irrelevant details. As a result, the model performs exceptionally well on the training data but poorly on new, unseen data, which means it fails to generalize. 

**`Underfitting`** occurs when a machine learning model is too simple to capture the underlying patterns in the training data. As a result, it performs poorly on both the training data and new, unseen data, indicating that it has not learned enough from the data.

**Key points:**

- `Overfitting` = High variance, low bias; fits training data too closely, poor generalization.
- `Underfitting` = High bias, low variance; too simple, poor fit on all data.
- Overfitting can be caused by complex models, small/noisy datasets, or overtraining.
- Underfitting can be caused by simple models, poor features, or insufficient training.
- The goal is to balance model complexity to achieve good performance on both training and unseen data.
- Techniques like cross-validation, regularization, and proper feature selection help mitigate both issues.
