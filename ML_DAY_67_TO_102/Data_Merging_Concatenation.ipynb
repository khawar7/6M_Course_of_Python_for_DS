# **Data Merging**

### **What is Data Merging?**
Data merging is the process of combining data from multiple sources into a single dataset. This can be done manually or using software tools. Data merging is commonly used in data analysis, data science, and business intelligence to create a unified view of data from different sources.

### **Why Is Data Merging Important?**

- `Holistic Analysis:` Merging data provides a complete picture, enabling deeper insights that would not be possible from isolated datasets.
- `Efficiency:` It reduces manual work by centralizing information, making it easier and faster to retrieve and analyze data.
- `Data Quality:` Helps in removing duplicates, correcting inconsistencies, and filling missing values, thereby improving data accuracy.
- `Informed Decision-Making:` A unified dataset supports better, data-driven decisions by presenting all relevant information together.

### **How Does Data Merging Work?**

The process typically involves several steps:

- `Data Profiling`: Understanding the structure and quality of each dataset.
- `Data Cleansing`: Identifying and correcting errors, inconsistencies, and duplicates.
- `Data Transformation`: Standardizing formats and structures to ensure compatibility.
- `Data Integration`: Combining the datasets, often using unique identifiers (like customer ID or product code).
- `Data Validation`: Checking the merged data for accuracy and completeness.

### **Common Techniques**

1. `Appending Rows:` Stacking datasets with the same structure (e.g., monthly sales reports) to increase the number of records.
2. `Appending Columns:` Adding new variables to existing records by matching on a common key (e.g., adding purchase history to customer demographics).
3. `Conditional Merge (Join):` Combining records based on specific conditions or keys, such as only merging rows where customer IDs match.
4. `Data Matching:` Identifying and combining similar or duplicate records from different sources, sometimes using fuzzy matching algorithms.

```python
# now we merge the three different datasets created on over own into a single one on row bases.

## 1. import the necessary libraries
import pandas as pd
import numpy as np

## 2. Create the three datasets

df1 = pd.DataFrame({                                   # dataset 1
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35]
})

df2 = pd.DataFrame({                                   # dataset 2     
    'ID': [2, 3, 4],
    'Name': ['Eva', 'Frank', 'Grace'],
    'Age': [28, 33, 40]
})

df3 = pd.DataFrame({                                   # dataset 3 
    'ID': [1, 4, 5],
    'Name': ['Alice', 'Grace', 'Hannah'],
    'Age': [25, 40, 45]
})

## 3. Merge the datasets
merged_df = pd.concat([df1, df2, df3], ignore_index=True)       

print(df1)
print("-------------------------------------------------------------")
print(df2)
print("-------------------------------------------------------------")
print(df3)
print("-------------------------------------------------------------")
print("----------------Merged Dataset on Row Bases------------------")
print(merged_df)
```

```python
# now we merge the three different datasets created on over own into a single one on Column bases.

## 1. import the necessary libraries
import pandas as pd
import numpy as np

## 2. Create the three datasets

df1 = pd.DataFrame({                                   # dataset 1
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35]
})

df2 = pd.DataFrame({                                   # dataset 2     
    'City': ["Istanbul","Ankara", "Izmir"],
    'House_No.': ['"28/A"', "33/B", "40/C"],
    'Block_No.': [8, 6, 4]
})
df3 = pd.DataFrame({                                   # dataset 3
'Profession': ['Engineer','Doctor','Artist'],
'Experience(yrs)': [15, 5, 10],
'Salary': ['250000$', '300000$', '3500000$']
})

## 3. Merge the datasets on columns and the names of colomns not ignored
merged_df_columns = pd.concat([df1, df2, df3], axis=1)

print(df1)
print("-------------------------------------------------------------")
print(df2)
print("-------------------------------------------------------------")
print(df3)
print("-------------------------------------------------------------")
print("----------------Merged Dataset on Column Bases----------------")
merged_df_columns
```

# **Concatenation of Data**

### **What Is Concatenation of Data?**

Concatenation of data is the process of linking two or more datasets or data elements together, either by adding rows (vertical concatenation) or columns (horizontal concatenation), to create a single, larger dataset. This operation is fundamental in data analysis and preparation, allowing analysts to combine information from different sources, add new records, or create new variables for analysis.

### **Why Is Data Concatenation Important?**

- `Data Integration`: Allows you to combine data from multiple sources into a unified dataset for comprehensive analysis.
- `Flexibility`: Enables you to append new records or variables as your data grows.
- `Efficiency`: Simplifies data management by reducing the number of separate files or tables.
- `Feature Creation`: Useful for creating new features by combining existing columns, especially in text or categorical data.

### **How Does Data Concatenation Work?**

The process of data concatenation generally involves:

- `Assessing Data Structure`: Ensuring datasets have compatible structures for concatenation (e.g., same columns for vertical, same rows for horizontal).
- `Selecting Concatenation Method`: Deciding whether to append rows or columns.
- `Performing Concatenation`: Using tools or functions (like concat in Python’s pandas or Excel’s CONCATENATE function) to combine the datasets.
- `Validating Results`: Checking the resulting dataset for accuracy, completeness, and consistency.

### **Common Techniques**

- `Vertical Concatenation`: Stacking datasets with the same columns to increase the number of records (e.g., adding monthly sales data).
- `Horizontal Concatenation`: Adding new columns to existing records, typically when datasets share the same number of rows (e.g., adding survey responses to demographic data).
- `String Concatenation`: Combining text fields to create new, meaningful values (e.g., merging first and last names into a full name).

```python
# now we concaterate the dataset created on over own

## 1. import the necessary libraries
import pandas as pd
import numpy as np

## 2. Create the dataset

df = pd.DataFrame({                                   # dataset
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35]
})


## 3. Concatenate the dataset
concatenated_df = pd.concat([df, df], ignore_index=True)  # Concatenate the same dataset
print(df)
print("-------------------------------------------------------------")
print("----------------Concatenated Dataset-------------------------")
print(concatenated_df)
```
