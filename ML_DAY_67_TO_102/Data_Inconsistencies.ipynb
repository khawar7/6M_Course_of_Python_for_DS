# **Data Inconsistencies or Anomolies**

`Definition`

Data inconsistencies and anomalies refer to unexpected, incorrect, or unusual data points that deviate from the standard, expected, or correct values in a dataset. These issues can arise due to errors in data collection, entry, transmission, or integration from multiple sources. Detecting and addressing them is crucial for accurate data analysis and modeling.

Common Types of Data Inconsistencies and Anomalies

1. `Missing Values`
- **Description**: Absence of data in one or more fields where a value is expected.
- **Causes:** Data entry errors, system failures, or incomplete data collection.
- **Example**: Name: Ali Khan | **Email Address: (blank)** | Phone: 0300-1234567


2. `Duplicate Records`
- **Description:** Multiple identical or near-identical records representing the same entity.
- **Causes:** Data merging from multiple sources, repeated data entry.
- **Example**: 
    1. Row 1: Name: Sara Ahmed | ID: 101 | Email: sara@example.com
    2. Row 2: Name: Sara Ahmed | ID: 101 | Email: sara@example.com


3. `Outliers`
- **Description**: Data points significantly different from the majority of the dataset.
- **Causes:** Measurement errors, rare events, or genuine variability.
- **Example**: Name: John Doe | Age: 200

4. `Inconsistent Data Formats`
- **Description**: Variations in how data is represented (e.g., date formats, units).
- **Causes**: Data integration from different systems or manual entry.
- **Example**: 
    1. Row 1: 2025-07-10
    2. Row 2: 10/07/2025
    3. Row 3: July 10, 2025

5. `Contradictory Data`
- **Description**: Conflicting information within or across records (e.g., age and birthdate mismatch).
- **Causes**: Data entry mistakes, merging errors.
- **Example**: Name: Ayesha Malik | Birthdate: 2010-01-01 | Age: 30

6. `Invalid Values`
- **Description**: Data entries that fall outside the allowed or logical range (e.g., negative ages, impossible dates).
- **Causes**: Input errors, lack of validation checks.
- **Example**: Product: Laptop | Quantity Sold: -5

7. `Structural Anomalies`
- **Description**: Unexpected changes in the structure or schema of the data (e.g., missing columns, extra fields).
- **Causes**: System updates, schema evolution, or integration issues.
- **Example**: 
    1. Row 1: Name, Age, Email
    2. Row 2: Name, Age
    3. Row 3: Name, Age, Email, Address


8. `Noise`
- **Description**: Random errors or irrelevant data that obscure the true signal.
- **Causes**: Sensor errors, transmission issues, or background processes.
- **Example**: Temperature readings: 22, 23, **150**, 22, 23


```python
# import the necessary libraries
import pandas as pd
```

```python
# now we create a dataset on over own that contain the four colomns and contain all the things that are being discussed in the file above
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Alice'],
    'Age': [25, 30, 350, 40],
    'Salary': [50000, 60, 70000, 80000],
    'Gender': ['F', 'NA', 'M', 'M'],
    'Date': ['01-01-2020', '05-15-2019', '2018-07-23', '2021-03-30'],
    'Country': ['U.S.A','USA','America','United States']
}
df=pd.DataFrame(data)
df
```

```python
# first we will deal with the name colomns that has a repeating name in it
df['Name'] = df['Name'].replace('Alice', 'Alicia')
df
```

```python
# Next is the age colomn we we see the age of 350 years that is not possible 
df['Age']=df['Age'].replace(350,35)
df
```

```python
# Now in the salary colomn we see the salary of 60 per month 
df['Salary']=df['Salary'].replace(60,60000)
df
```

```python
# In the Gender colomn we have so one missing value of NA value in it
df['Gender']=df['Gender'].replace('NA','M')
df
```

```python
# The country colomn also has some inconsistency so we will fix it too all at once
df['Country']=df['Country'].replace('USA','United States')
df['Country']=df['Country'].replace('U.S.A','United States')
df['Country']=df['Country'].replace('America','United States')
df
```

```python
# standardize Date column to ISO format (YYYY-MM-DD)
df['Date'] = pd.to_datetime(df['Date'], infer_datetime_format=True, errors='coerce')
df['Date'] = df['Date'].dt.strftime('%Y-%m-%d')
df
```

```python
# now we see the case of missing values in date colomns
df['Date']=df['Date'].replace('NaT','23-007-2018')
df['Date']=df['Date'].replace('')
```
