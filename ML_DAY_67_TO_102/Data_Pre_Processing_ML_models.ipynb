# **Data Pre-Processing Beofore doing the Machine Learning Model**

Data pre-processing is a `critical and foundational step for successful machine learning model development`. It prepares raw data, making it usable and optimal for machine learning models. Data scientists and machine learning engineers typically spend `80% of their time on data pre-processing`, with only `20% dedicated to model development`. If this crucial 80% is not handled correctly, the entire project can fail, leading to issues with stakeholders or clients.

Here are the detailed aspects of data pre-processing discussed in the video:

*   `Understanding Data Origin and Purpose`
    *   Before working with any data, it is essential to be **aware of where the data comes from** and what procedures can be applied to it.
    *   Data collection is driven by a **specific aim or question**. This aim guides the collection process, determining what data is acquired.
    *   `Metadata` (data about data) is as important as the data itself. It includes information such as who collected the data, the purpose of its collection, and the number of columns.
    *   Assuming that data is 100% correct, even if collected by a knowledgeable person, is a **mistake**. Data is rarely perfect, and processes are needed to handle it.

*   `The Necessity of Data Pre-processing`
    *   Data pre-processing means **making the data suitable for use by a machine learning model**, aiming for up to 100% correctness for the model's purposes.
    *   It is `always needed`, even if the data appears clean, because certain concepts within pre-processing are necessary to improve the data and enable the machine learning model to work effectively.
    *   Neglecting pre-processing can lead to the model not performing correctly, making the entire effort futile.

*   `Key Steps in Data Pre-processing`
    The process of data pre-processing involves five crucial steps:

    1.  **Data Cleaning**
        *   `Purpose:` To make the data tidy, well-maintained, and clean.
        *   `Tidying Data:` Organizing data of one type together, and separating different types, similar to organizing a room.
        *   `Handling Noisy Data and Outliers:` Removing irrelevant or erroneous information, which can be thought of as "shoppers" on a shelf that serve no useful purpose. These can be discarded or sometimes maintained.
        *   `Filling Missing Values (Imputation):` Addressing gaps in the data, similar to filling empty hangers in a wardrobe with clothes.
        *   `Resolving Inconsistencies:` Ensuring data is structured uniformly and that elements are in their correct places, preventing mixing of unrelated data (e.g., kitchen items in a clothes shelf).
        *   While not strictly cleaning, sometimes new data columns or variables can be created based on existing ones, like creating a 'months' column from an 'age in years' column, which falls under feature engineering.

    2.  **Data Integration**
        *   `Purpose:` Combining data collected from multiple methods or sources (e.g., via **triangulation**, where the same task is performed using multiple methods to arrive at a single point). This is also known as **data pooling**.
        *   `Examples:` Combining data from multiple people sent to count male and female students in different universities in a city.
        *   `Challenges:` Data integration often introduces **redundancy** and **duplicates** (data appearing multiple times).
        *   `Post-Integration Cleaning:` After integrating data, it is often necessary to perform **another round of data cleaning** to remove new duplicates and inconsistencies that arise from the integration process.

    3.  **Data Transformation**
        *   `Purpose:` Transforming data into a form suitable for analysis and modeling, often bringing it to a consistent scale or combining information.
        *   `Data Scaling and Normalization:` Adjusting numerical data to a common scale to prevent features with larger values from dominating the model.
        *   `Data Aggregation:` Combining two or more variables (features) into a single, more meaningful variable.
            *   `Example:` In Titanic data, combining 'sibling/spouse presence', 'alone travel', and 'parent/children presence' columns to create a single 'Family Size' column.
            *   `Variables, Features, and Labels:` All columns in a dataset are called **variables**. Variables can be either **features** (inputs, e.g., x1, x2, x3) or a **label** (output, e.g., y).
        *   `Data Generalization:` Transforming low-level, specific data into broader, higher-level concepts.
            *   `Example:` Binning specific 'age' values into age groups (e.g., 5 years old, 5-10 years, 11-18 years, etc.).
            *   This process is called **concept hierarchical distribution** and makes the data more general for machine learning models.
            *   `Example from Titanic:` The 'Survived' column is often used as the output (label) to predict survival based on other features like age, fare, class, and gender.

    4.  **Data Reduction**
        *   `Purpose:` Reducing the volume of data while retaining its important information, which is crucial for computational efficiency and model performance.
        *   `Dimensionality Reduction:` Reducing the number of input features (columns) in the dataset.
            *   `Reasoning:` Not all features are equally important for predicting the output; some may have less impact. For example, when traveling, features like destination, mode of transport, and money are important, but whether one ate food might not be directly relevant to reaching the destination.
            *   `Benefits:` Reduces the data's "dimensions," saving computational resources and focusing on the most impactful features.
            *   `Techniques:` Principal Component Analysis (PCA), multivariate analysis, Gosch mixture models, and t-SNE are used for this.
        *   `Numerosity Reduction:` Converting categorical text data into numerical representations.
            *   `Example:` Replacing "Male" and "Female" with "1" and "0" respectively.
            *   `Benefits:` Reduces the number of characters/alphabets, saving storage space, and speeds up processing because computers process numbers faster than alphabets.
            *   `Techniques:` Data encoding techniques like One-Hot Encoding and Label Encoding.
        *   `Data Compression:` Using various methods to compress the dataset for easier storage and transfer.
        *   `Context:` Data reduction is especially important for those with limited computational resources and for pilot projects.

    5.  **Data Discretization**
        *   `Purpose:` Converting continuous numeric data into discrete (countable) nominal or categorical variables.
        *   `Example:` Converting exact ages into age groups (e.g., 5-10 years, 11-18 years). This is similar to data generalization but specifically focuses on converting numeric to categorical.
        *   `Application:` Can be used with binning algorithms. For example, converting exact fare prices into categories like "cheap," "medium," or "expensive".
        *   `Benefits:` Simplifies data representation and can be very useful when converting numeric variables into categorical ones for certain models or analyses.
