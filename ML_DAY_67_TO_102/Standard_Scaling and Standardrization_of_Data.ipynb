# **Standard Scaling and Standardization of Data**

### **1. What is Standard Scaling?**

`Standard scaling` (also called standardization or z-score normalization) is a data preprocessing technique that transforms numerical features so that they have a mean of 0 and a standard deviation of 1. This process is essential in data analysis and machine learning to ensure that features with different units or scales contribute equally to model training.

How Does Standard Scaling Work?
The transformation is performed using the formula:

X_scaled = X − μ / σ. 

Where:

- X = original value
- μ = mean of the feature
- σ = standard deviation of the feature

This formula centers the data around zero and scales it so that the spread (variance) is consistent across features.

### **Why Use Standard Scaling?**

- `Equal Contribution`: Ensures all features contribute equally to model training, especially important for algorithms sensitive to feature scale (e.g., SVM, logistic regression, PCA).
- `Improved Model Performance`: Helps gradient-based algorithms converge faster and prevents features with larger scales from dominating the learning process.
- `Outlier Handling`: Reduces the impact of outliers, though it is still somewhat sensitive to extreme values.

```python
# import the necessary libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
## 1. Create a exmple DataFrame
data = {
    'age': [25, 30, 35, 40, 45],
    'income': [50000, 60000, 70000, 80000, 90000],
    'height': [150, 160, 170, 180, 190]
}

df = pd.DataFrame(data)
print(df)
```

```python
## 2. import the StandardScaler
scaler = StandardScaler()

## 3. Fit the scaler to the DataFrame
scaled_data = scaler.fit_transform(df)
scaled_data
```

```python
## convert the scaled data back to a DataFrame
scaled_df = pd.DataFrame(scaled_data, columns=df.columns)
print(scaled_df)
```

### **2. Min_Max Scalar**
