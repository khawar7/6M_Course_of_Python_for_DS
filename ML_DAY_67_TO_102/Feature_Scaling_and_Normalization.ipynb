# **Scaling and Normalization of Data**

### **What is scaling and normalization of data?**

Scaling and normalization of data are techniques used to transform the data into a common scale, making it easier to compare and analyze. Scaling involves adjusting the range of values in the data, while normalization involves transforming the data into a specific range, usually between 0 and 1.

### **What is the purpose of scaling and normalization?**

The purpose of scaling and normalization is to ensure that all features in the data have the same scale, making it easier to compare and analyze the data. This is particularly important when using machine learning algorithms, as they often require the data to be on the same scale.
 
### **What are the benefits of scaling and normalization?**

The benefits of scaling and normalization include:

*   Ensuring that all features have the same scale, making it easier to compare and analyze the data.
*   Improving the performance of machine learning algorithms.
*   Reducing the impact of outliers and noisy data.
*   Enhancing the interpretability of the results.
*   Improving the robustness of the models.
*   Reducing the impact of features with large ranges of values.
*   Improving the performance of machine learning algorithms.

*   **Why Data Transformation is Important**
    *   **Normal Distribution and Machine Learning Models**: Most machine learning models perform well if the **data is normally distributed** (i.e., forms a bell curve with mean at the center, normal standard deviation, kurtosis, and skewness).
    *   **Handling Skewed Data**: Often, data is not normally distributed; it can be **right-skewed**, **left-skewed**, or **bimodal**. Data transformation helps in normalizing such distributions.
    *   **Outlier Impact**: Outliers can **skew or distort the data**, pulling the distribution out of its normal shape.
    *   **Outlier Removal**: Removing outliers can help straighten the data.
    *   **Scaling for Outliers**: When outliers have very large values (e.g., 80 in a 1-5 range), **standardizing the entire data** (bringing it to a smaller scale) can be beneficial.
    *   **Computational Knowledge**: We scale values to make them smaller due to **computational knowledge**.
    *   **Maintaining Differences**: When data is transformed or scaled, the **inherent differences, dimensions, and magnitude between data points remain the same**; only their values become smaller. For instance, if the difference between 1 and 80 is significant, the difference between 0.1 and 0.8 will also be proportionally significant.
    *   **Increased Chances of Normal Distribution**: Data transformation **increases the chances of the data achieving a normal distribution**.

*   **Concept of Data Transformation Defined**
    *   Transformation means converting data from its **original form to a new form** that is **considerably important and easy to use**.
    *   This results in **modified data** or **transformed data**.
    *   This process involves applying a **procedure or algorithm** (also called a **transformer**).

*   **Types of Linear Data Transformation (Scalers)**
    *   **Standardization (Standard Scaler)**
        *   Goal: To **standardize the data** by applying an equation to bring it into a standard format.
        *   Common Range: **Usually -3 to +3**. This range is related to the Z-score, which ranges from -3 to +3.
        *   Use Case: Used when the **machine learning algorithm can handle negative values**.
    *   **Min-Max Scaler (Scaling to a Non-Range)**
        *   Goal: To **scale data to a specific non-range**, typically **0 to 1**.
        *   Name in Scikit-learn: **MinMaxScaler** (minimum 0, maximum 1).
        *   Use Case: Used when the **machine learning algorithm cannot handle negative values**.
    *   **Max Absolute Scaler**
        *   Goal: To scale data to a range of **-1 to +1**.
        *   Name: **MaxAbsoluteScaler**.
        *   Use Case: Used when the **machine learning algorithm can handle negative values**.
    *   **Outlier Handling with Scalers**: Some scalers can **automatically deal with outliers** by scaling them along with the rest of the data.

*   **Non-Linear Transformations**
    *   Purpose: These transformers directly convert your data into a **Gaussian (normal) distribution** or **Uniform distribution**.
    *   **Quantile Transformer**
        *   Function: Performs **non-parametric transformation** of data.
        *   Output: Converts data into a **uniform transformation**.
        *   Can also transform data into a **normal distribution** (often for heavy transformations). For example, it can convert Uniform or Bimodal distributions into Gaussian distributions.
    *   **Mapping to Gaussian Distribution**
        *   Methods are available to map non-normal data distributions (e.g., **log-normal, chi-squared, Weibull distributions**) to a **Gaussian distribution**.
        *   **Box-Cox Transformer**
            *   Converts **log-normal, chi-squared, and Weibull distributions to a normal distribution**.
            *   Effect on other distributions: Applying Box-Cox to **Gaussian data results in no significant change** (as it's already Gaussian). For **Uniform and Bimodal data, Box-Cox often shows no significant change**.
        *   **Yeo-Johnson Transformer**
            *   Also transforms distributions to a **normal distribution**.
            *   Similar to Box-Cox, applying Yeo-Johnson to **Gaussian, Uniform, or Bimodal distributions often results in no significant change**.

*   **Normalization vs. Standard Scaling/Scaling**
    *   **Normalization**
        *   Process: **Scaling individual samples to have a unit norm**.
        *   Usefulness: Can be useful if you plan to use a **quadratic form** (like a dot product or any other kernel to quantify similarity between samples).
        *   Effect: **Individual values become smaller**.
        *   Example: **Log transformation** (log 2, log 10, or simple log) can normalize data by converting large quantities into smaller ones. This is not a range-based scaling but rather a value-reducing transformation.
    *   **Standard Scaling/Scaling**
        *   Process: Involves **scaling the entire dataset** to a particular range or standard.
