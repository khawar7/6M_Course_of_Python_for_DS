# **Matrix  or Matrices**

```python
# import the necessay library
import numpy as np
```

```python
# now we create the 2 x 3 matrix 
a=np.array([[1,2,3],[4,5,6]])
print(a)
```

```python
# create a matrix of big dimension
b=np.arange(0,100,2)
print(b)
```

```python
print(a[1])
```

```python
print(b[1])
```

```python
b_shape=b.reshape(5,10)
print(b_shape)
```

```python
# slicing the matrix
b_shape[1:3,2:4]
```

## **Now we see the min , max and the sum of the values in the matrices**

```python
b_shape.min()
```

```python
b_shape.max()
```

```python
b_shape.sum()
```

```python
b_shape.min(axis=0)     # min of each column
```

```python
b_shape.min(axis=1)     # min of each row
```

```python
b_shape.max(axis=0)     # max of each column
```

```python
b_shape.max(axis=1)     # max of each row
```

### **Add numbers in Rows or Colomns in the Martix**

```python
# now we create a new array using the arange function
c=np.arange(10,50,2).reshape(5,4)
print(c)
```

```python
c+np.array([50,52,54,56])
```

```python
# Add a new row
c=np.append(c,[[10,12,14,16]],axis=0)
print(c)
```

```python
# Add a new column to matrix c
c = np.append(c, [[18], [20], [22], [24], [26], [28]], axis=1)
print(c)
```

```python
# Reverse an array
c[::-1]
```
